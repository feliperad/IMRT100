```python
''' In Python, almost all variables store references to the actual data. 
When such variables are copied or passed to a function, they copy the reference, not the actual data'''

a = [1, 2]
b = a
b[0] = 55

print(a)

b = a.copy()
b[0] = 55

print(f'a is {a} and b is {b}')
```

### For loop

```python
for i in range(5):
    print('The value is',i)

print('\n')

list = [34, 53, 532, 23, 1]

for i in list:
    print(i)

print('\n')

for i, item in enumerate(list):
    print(i)
    print(item)

print('\n')

for i, item in enumerate(list, start=1):
    print(f'item number {i}: {item}')

print('\n')

even_numbers = [number for number in list if number%2==0]
print('the even numbers of the list are', even_numbers)
```

### While loop

```python
b = 0

while b<10:
    print(b)
    b+=1

print('\n')

c = 0
while True:
    print(c)
    c+=1
    if c >= 10:
        break

```

### Creating functions

```python
def sum_of_three_numbers(n1,n2,n3):
    return n1+n2+n3

sum_value = sum_of_three_numbers(2,5,5)
print("the sum value is", sum_value)

print('\n')

def number_check(list,threshold=5):
    '''check if number is odd and greater than the threshold'''

    for number in list:
        if number%2==0:
            continue 

        if number>threshold:
            print(f'number {number} is odd and greater than {threshold}')
            continue

        print(f'number {number} is odd but smaller than or equal to {threshold}')


number_check([1, 3, 5, 6, 10],2)
print('\n')
number_check([1, 3, 5, 6, 10])
```
